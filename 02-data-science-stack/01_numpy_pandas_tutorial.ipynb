# NumPy and Pandas for Machine Learning

This notebook introduces core NumPy array operations and Pandas DataFrame workflows commonly used in machine learning.

```python
import numpy as np
import pandas as pd

np.set_printoptions(suppress=True)
pd.set_option('display.max_columns', 20)
pd.set_option('display.precision', 2)
```

## Array creation in NumPy

NumPy provides fast constructors for common patterns such as zeros, ones, ranges, and random samples.

```python
rng = np.random.default_rng(42)

zeros = np.zeros((2, 3))
ones = np.ones((2, 3))
sequence = np.arange(0, 12, 2)
grid = np.linspace(0, 1, 5)
random_values = rng.normal(loc=0.0, scale=1.0, size=(3, 3))

print('zeros:')
print(zeros)
print('ones:')
print(ones)
print('sequence:', sequence)
print('grid:', grid)
print('random_values:')
print(random_values)
```

## Indexing, slicing, and filtering

Slicing extracts structured subsets, while fancy indexing and boolean masks help us select specific rows or values.

```python
X = np.arange(1, 13).reshape(3, 4)
print('X:')
print(X)
print('Second row:', X[1])
print('Last two columns:')
print(X[:, 2:])
print('Fancy row selection:')
print(X[[0, 2]])

mask = X % 2 == 0
print('Even entries:', X[mask])
```

## Element-wise operations and broadcasting

NumPy applies arithmetic element-wise and can broadcast smaller arrays across larger ones.

```python
scores = np.array([[78, 85, 90], [88, 79, 92], [95, 91, 87]])
bonus = np.array([2, 0, 3])

adjusted = scores + bonus
centered = scores - scores.mean(axis=0)
scaled = centered / scores.std(axis=0)

print('Adjusted scores:')
print(adjusted)
print('Centered scores:')
print(centered)
print('Scaled scores:')
print(scaled)
```

## Reshaping and stacking

Reshaping reorganizes values without changing them. Stacking combines arrays into wider or taller structures.

```python
flat = np.arange(12)
matrix = flat.reshape(3, 4)
column = np.array([100, 200, 300]).reshape(-1, 1)

print('Matrix:')
print(matrix)
print('Flattened:', matrix.ravel())
print('Column stack:')
print(np.hstack([matrix[:, :1], column]))
print('Vertical stack:')
print(np.vstack([matrix[:1], matrix[2:]]))
```

## Linear algebra tools

Many machine learning algorithms depend on matrix multiplication and decompositions.

```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
B = np.array([[1.0, 4.0], [2.0, 5.0]])

print('dot:')
print(np.dot(A, B))
print('matmul:')
print(A @ B)
print('inverse of A:')
print(np.linalg.inv(A))
print('determinant of A:', np.linalg.det(A))
values, vectors = np.linalg.eig(A)
print('eigenvalues:', values)
print('eigenvectors:')
print(vectors)
```

## Creating labeled data with Pandas

Pandas adds row and column labels, which makes tabular data easier to inspect and manipulate.

```python
students = pd.Series([88, 93, 79], index=['Alice', 'Ben', 'Chloe'], name='quiz_score')

df = pd.DataFrame({
    'student': ['Alice', 'Ben', 'Chloe', 'Darius', 'Ella'],
    'study_hours': [6.5, 8.0, 4.0, 7.5, np.nan],
    'attendance': [0.95, 0.90, 0.80, 0.88, 0.92],
    'passed': [True, True, False, True, True]
})

print(students)
df
```

```python
print(df.loc[0:2, ['student', 'study_hours']])
print(df.iloc[:3, :2])
print(df[df['attendance'] >= 0.9])
```

```python
messy_df = pd.concat([df, df.iloc[[1]]], ignore_index=True)
messy_df.loc[4, 'study_hours'] = np.nan

cleaned_df = (
    messy_df
    .drop_duplicates()
    .assign(study_hours=lambda frame: frame['study_hours'].fillna(frame['study_hours'].median()))
)

print('Rows before cleaning:', len(messy_df))
print('Rows after cleaning:', len(cleaned_df))
cleaned_df
```

## Aggregation and joining

GroupBy summarizes data by category, and merges combine related tables from different sources.

```python
summary = cleaned_df.groupby('passed').agg(
    avg_hours=('study_hours', 'mean'),
    avg_attendance=('attendance', 'mean'),
    count=('student', 'count')
)
summary
```

```python
advisor_df = pd.DataFrame({
    'student': ['Alice', 'Ben', 'Chloe', 'Darius', 'Ella'],
    'mentor': ['Dr. Lee', 'Dr. Lee', 'Dr. Kim', 'Dr. Kim', 'Dr. Singh']
})

merged = cleaned_df.merge(advisor_df, on='student', how='left')
merged
```
