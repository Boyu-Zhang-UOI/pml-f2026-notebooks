# Python Basics for Machine Learning
A compact tutorial on Python fundamentals used throughout the course.


## 1. Data types and variables
Python variables are dynamically typed, which makes experimentation fast and flexible.


```python
examples = 150
learning_rate = 0.01
model_name = 'linear regression'
is_ready = True
missing_value = None
print(type(examples), type(learning_rate), type(model_name), type(is_ready), type(missing_value))
```

```python
train_accuracy = 0.93
message = f"{model_name.title()} reached {train_accuracy:.1%} accuracy"
print(message)
```

## 2. Strings and type conversion
Real-world data often arrives as text, so conversions are common.


```python
raw_rows = '250'
raw_score = '0.875'
rows = int(raw_rows)
score = float(raw_score)
print(rows + 50)
print(f"Validation score: {score:.2%}")
```

## 3. Lists and dictionaries
These structures appear constantly in feature engineering and experiment tracking.


```python
scores = [0.76, 0.81, 0.93, 0.88]
scores.append(0.91)
print(scores[1:4])
summary = {'model': 'tree', 'max_depth': 4, 'score': max(scores)}
print(summary)
```

```python
high_scores = [value for value in scores if value >= 0.9]
score_map = {i: value for i, value in enumerate(scores, start=1)}
print(high_scores)
print(score_map)
```

## 4. Conditionals
Boolean expressions help programs decide what to do next.


```python
score = 0.88
if score >= 0.9:
    print('Excellent')
elif score >= 0.8:
    print('Good')
else:
    print('Needs improvement')
```

## 5. Loops
Loops let us process collections one item at a time.


```python
for epoch in range(1, 4):
    print(f'Epoch {epoch}')

countdown = 3
while countdown > 0:
    print('Countdown:', countdown)
    countdown -= 1
```

## 6. Functions
Functions make code reusable and easier to test.


```python
def normalize(values):
    total = sum(values)
    return [value / total for value in values]

def describe_run(model, **kwargs):
    return {'model': model, **kwargs}

print(normalize([2, 3, 5]))
print(describe_run('svm', kernel='rbf', c=1.0))
```

```python
pairs = [('linear', 0.84), ('forest', 0.91), ('svm', 0.89)]
pairs.sort(key=lambda item: item[1], reverse=True)
print(pairs)
```

## 7. Generators
Generators yield values one at a time, which is helpful for large datasets.


```python
def running_total(values):
    total = 0
    for value in values:
        total += value
        yield total

for partial in running_total([1, 2, 3, 4]):
    print(partial)
```

## 8. Mini class example
Many machine learning libraries are built around classes and objects.


```python
class Experiment:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def best_score(self):
        return max(self.scores)

    def __repr__(self):
        return f"Experiment(name={self.name!r}, scores={self.scores!r})"

trial = Experiment('baseline', [0.76, 0.81, 0.88])
print(trial)
print('Best:', trial.best_score())
```
